```python

description = input("Enter report description: ").lower()
audience = input("Enter intended audience: ").lower()

text = description + " " + audience

if any(word in text for word in
       ["executive", "management", "business risk", "long-term", "strategy"]):
    print("Classification: Strategic")

elif any(word in text for word in
         ["tactic", "technique", "attacker method", "mitre", "security team"]):
    print("Classification: Tactical")

elif any(word in text for word in
         ["ip address", "ioc", "malicious domain", "incident", "soc analyst"]):
    print("Classification: Operational")

else:
    print("Classification: Unable to classify")

```
