```python

connections = set()

while True:
    print("\n1. Outbound connection")
    print("2. Inbound packet")
    print("3. Exit")

    choice = input("Enter choice: ")

    if choice == "1":
        connection = input("Enter connection ID: ")
        connections.add(connection)
        print("Outbound connection allowed.")
        print("Connection tracked by firewall.")

    elif choice == "2":
        connection = input("Enter connection ID: ")

        if connection in connections:
            print("Inbound packet allowed.")
            print("Return traffic belongs to an established connection.")
        else:
            print("Inbound packet BLOCKED.")
            print("Unsolicited inbound traffic detected.")

    elif choice == "3":
        print("Firewall simulation ended.")
        break

    else:
        print("Invalid choice.")
```
