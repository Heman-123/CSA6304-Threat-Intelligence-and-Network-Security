```python

activity = input("Describe intelligence team activity: ").lower()

if any(word in activity for word in
       ["requirement", "objective", "priority", "planning"]):
    print("Stage: Direction")

elif any(word in activity for word in
         ["gather", "collect", "obtain", "sources"]):
    print("Stage: Collection")

elif any(word in activity for word in
         ["clean", "normalize", "filter", "organize"]):
    print("Stage: Processing")

elif any(word in activity for word in
         ["analyze", "correlate", "interpret", "investigate"]):
    print("Stage: Analysis")

elif any(word in activity for word in
         ["share", "distribute", "deliver", "report"]):
    print("Stage: Dissemination")

elif any(word in activity for word in
         ["feedback", "review response", "improve requirements"]):
    print("Stage: Feedback")

else:
    print("Stage: Unable to identify")
```
