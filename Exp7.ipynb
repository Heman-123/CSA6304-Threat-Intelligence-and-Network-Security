```python

import json

logs = [
    'INFO user=alice status=success',
    'ERROR user=bob status=failed',
    'INFO user=alice status=success',
    'ERROR user=carol status=failed'
]

# Logstash-style parsing
events = []

for log in logs:
    parts = log.split()
    event = {
        "level": parts[0],
        "user": parts[1].split("=")[1],
        "status": parts[2].split("=")[1]
    }
    events.append(event)

print("Parsed events:")
for event in events:
    print(event)

# Elasticsearch-style indexing
index = {}

for i, event in enumerate(events):
    index[i] = event

print("\nIndexed events:", len(index))

# Kibana-style aggregation
counts = {}

for event in index.values():
    status = event["status"]
    counts[status] = counts.get(status, 0) + 1

print("\nStatus summary:")
for status, count in counts.items():
    print(status, ":", count)
```
