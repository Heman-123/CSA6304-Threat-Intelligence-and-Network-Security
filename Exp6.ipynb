```python

from urllib.parse import urlparse
import ipaddress

url = input("Enter URL: ").strip()
check_url = url if "://" in url else "https://" + url
parts = urlparse(check_url)

host = parts.hostname or ""
flags = []

keywords = ["login", "verify", "update", "banking", "password"]

for word in keywords:
    if word in url.lower():
        flags.append("Suspicious keyword: " + word)

try:
    ipaddress.ip_address(host)
    flags.append("IP address used as domain")
except ValueError:
    pass

if len(host.split(".")) > 4:
    flags.append("Excessive subdomains")

if parts.scheme != "https":
    flags.append("Not using HTTPS")

if flags:
    print("Warning: Suspicious URL indicators found")
    for flag in flags:
        print("-", flag)
else:
    print("No listed phishing indicators found")
```
