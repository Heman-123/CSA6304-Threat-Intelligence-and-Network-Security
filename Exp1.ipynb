```python

scenario = input("Enter business scenario: ").lower()

if "bank" in scenario or "financial" in scenario or "fraud" in scenario:
    print("Application: Banking Fraud Prevention")
elif "hospital" in scenario or "healthcare" in scenario or "patient" in scenario:
    print("Application: Healthcare Data Protection")
elif "e-commerce" in scenario or "online shopping" in scenario or "retail" in scenario:
    print("Application: E-commerce Fraud Prevention")
else:
    print("Application: General Threat Intelligence")
```
