```python

policy = input("Enter security policy text: ").lower()

elements = {
    "Scope": ["scope", "applies to"],
    "Roles and Responsibilities": ["roles", "responsibilities"],
    "Enforcement": ["enforcement", "disciplinary action"],
    "Review Cycle": ["review cycle", "review annually"],
    "Approval Authority": ["approved by", "approval authority"]
}

missing = []

for element, keywords in elements.items():
    found = False

    for word in keywords:
        if word in policy:
            found = True

    if not found:
        missing.append(element)

if len(missing) == 0:
    print("All governance elements are present.")
else:
    print("Missing governance elements:")
    for item in missing:
        print("-", item)

```
